### Importation des packages necessaires

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from tqdm import tqdm
```

### Nous commencons par charger notre image couleur RGB

```python
def charger_image(chemin_image):
    
    image = Image.open(chemin_image)
    image = image.convert('RGB')  # S'assurer que c'est couleur
    return np.array(image)
```

### Ici nous transformons notre image RGB de dimension trois (3D) ent dimension deux (2D) pour cela nous applicons une combinaison lineaire en ponderand le canal RGB avec des coefficients standards en gros tout ce passe comme si on applatie un cube sur une de ses dimension

```python
def convertir_en_gris(image_rgb):
    """
    Convertit une image RGB en image 2D niveaux de gris.
    Utilisation de la formule standard : 0.299*R + 0.587*G + 0.114*B
    """
    rouge = image_rgb[:, :, 0]
    vert  = image_rgb[:, :, 1]
    bleu  = image_rgb[:, :, 2]

    image_gris = 0.299*rouge + 0.587*vert + 0.114*bleu
    return image_gris.astype(np.uint8)
```

### Une fois que notre image est en niveau de gris, on peut appliquer la Singular Value Decomposition.
### La compression SVD : Singular Value Decomposition est C’est une technique mathématique pour décomposer une matrice (comme une image) en 3 parties :
### M = U⋅S⋅V.T où M est matrice originale (image), U le vecteurs “de base” dans la direction des lignes, S la valeurs singulières (importance de chaque vecteur de base) et Vᵀ vecteurs “de base” dans la direction des colonnes

k = nombre de valeurs singulières à conserver, c’est-à-dire le nombre de “composantes importantes” qu’on garde pour reconstruire l’image


```python
def reconstruire_image_svd(image_2d, k):

    # Applique la SVD sur une image 2D et ne garde que k valeurs singulières.
    
    U, S, Vt = np.linalg.svd(image_2d, full_matrices = False)
    
    S_k = np.zeros((k, k))
    np.fill_diagonal(S_k, S[:k])
    U_k = U[:, :k]
    Vt_k = Vt[:k, :]
    
    image_compressee = U_k @ S_k @ Vt_k
    image_compressee = np.clip(image_compressee, 0, 255).astype(np.uint8)
    
    return image_compressee
```

### Après compression de notre image, il est necessaire de connaitre deux parametres important qui sont :ratio_compression qui montre combien la taille de l’image a été réduite l'erreur qui informe sur combien l’image compressée diffère de l’originale. 

```python
def calculer_metrics_2d(image_originale, image_compressee, k):
    hauteur, largeur = image_originale.shape

    taille_originale = hauteur * largeur
    taille_compressee = k * (hauteur + largeur + 1)

    ratio_compression = taille_compressee / taille_originale
    erreur = np.linalg.norm(image_originale - image_compressee)

    return ratio_compression, erreur
```

### Exemple d'utilisation pas à pas

```python
# Charger l’image RGB
chemin = 'Images/image.jpg'
image_rgb = charger_image(chemin)

# Convertir en niveaux de gris
image_gris = convertir_en_gris(image_rgb)

# image avant compression
plt.imshow(image_gris, cmap = 'gray')
plt.title("image sans compression")
plt.axis('off')

# Définir les valeurs de k à tester
valeurs_k = [5, 20, 50, 100]

plt.figure(figsize=(12, 8))

#  Boucle pour compresser et afficher
from tqdm import tqdm

# Boucle avec barre de progression
for i, k in enumerate(tqdm(valeurs_k, desc="Compression SVD"), 1):
    image_compressee = reconstruire_image_svd(image_gris, k)
    ratio, erreur = calculer_metrics_2d(image_gris, image_compressee, k)

    plt.subplot(2, 2, i)
    plt.imshow(image_compressee, cmap='gray')
    plt.title(f'k={k}, compression={ratio:.2f}, erreur={erreur:.2f}')
    plt.axis('off')

plt.tight_layout()
plt.show()

```
